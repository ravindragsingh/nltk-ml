```python
(x_train,y_train),(x_test,y_test) = keras.datasets.mnist.load_data()

```

```python
x_train=x_train/255
x_test = x_test/255
#scaling the values from larger to smaller value
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape =(28,28)),
    keras.layers.Dense(1000,activation ='relu'),#rectified linear unit for all the positvie value it gives same value for all the neagtive value it gives 0
    keras.layers.Dense(200,activation ='relu'),
    keras.layers.Dense(10,activation ='softmax')#multiple category we use softmax and two category we use sigmoid

])
#f(x) = max(0,x)
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=2)
```

```python
60000/1875
```

```python
model.evaluate(x_test,y_test)
```

```python
x_train = x_train.reshape(-1,28*28)
x_test = x_test.reshape(-1,28*28)
```

```python
print(x_train[0])#all deeplearning model takes input in single dim array
```

```python
model = keras.Sequential([
    keras.layers.Dense(1000,activation ='relu',input_shape=(784,)),#rectified linear unit for all the positvie value it gives same value for all the neagtive value it gives 0
    keras.layers.Dense(500,activation ='relu'),
    keras.layers.Dense(200,activation ='relu'),
    keras.layers.Dense(10,activation ='softmax')#multiple category we use softmax and two category we use sigmoid

])
#f(x) = max(0,x)
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=2)
```

```python
model.evaluate(x_test,y_test)
```
